```python
#Fórmula de Herón

a = 7
b = 8
c = 9

# a + b > c
# a + c > b
# b + c > a
# Las anteriores condiciones se deben de cumplir para que sea un triángulo

s = (a + b + c) / 2

area = (s*(s-a)*(s-b)*(s-c)) ** 0.5

print(f"El semiperimetro es: {s} ")
print(f"El área es: {area} ")
```
